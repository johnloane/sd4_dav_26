All note to follow come from The Visual Display of Quantitative Information by Edward Tufte.

Excellence in statistical graphics consists of complex ideas communicated ith clarity, efficiency, and precision. Graphical displays should:
- Show the data
- Induce the viewer to think about the substance of the data rather than about methodology, graphic design, the technology used to produce the graph of anything else
- Avoid distorting what the data has to say
- Present many numbers in a small space
- Make large datasets coherent
- Encourage the eye to compare different pieces of data
- Reveal the data at several levels of detail, from broad view to a fine structure
- Serve a reasonably clear purpose; description, exploration, tabulation
- Be closely integrated with the statistical and verbal descriptions of the dataset
- Graphics should reveal data

**Graphics Integrity**
- Lie factor = $\frac{size of the effect shown in the graphic}{size of the effect in the data}$
- The representation of numbers, as physcially measured on the surface of the graphic itself, should be directly proportional to the numerical quantities present
- Clear, detailed, and thorough labelling should be used to defeat graphical distortion and ambiguity. Write out explanation of the data on the graphic. Label important events in the data
- Show data variation, not design variation
- In time-series displays of money, deflated and standardised units of money are nearly always better
- The number of information carrying variables dimensions, depicted should not exceed the number of dimensions in the data
- Graphics must not quote data out of context

**Sources of Graphical Integerity and Sophistication**
- Lack of quantitative skills in people producing graphs
- The idea that statistical data is bring
- If statistics are boring you have the wrong numbers
- The doctrine that graphics are only for the unsophisticated reader
- Aimed to divert and entertain those who find the text too difficult
- E.B. White "No one can write decently who is distrustful of the reader's intelligence or whose attitude is patronising
- Contempt for graphics and their audience, along with a lack of quantitative skills, has deadly consequences for graphics work: over decorated and simplistic designs, tiny data sets, and big lies

**Theory of Data Graphics - Data-Ink and Graphical Redesign**
- Data graphics should draw the viewer's attention to the sense and substance of the data, not something else
- The data graphics should present the quantitative contents
- Statistical graphics are instruments to help people reason about quantitative information
- The fundamental principle of good statistical graphics: Above all else show the data

**Data ink**
- A large share of ink on a graphic should present data information, the ink changing as the data changes
- Data-ink is the non-erasable core of a graphic, non-redundant ink arranged in response to variation in the numbers presented
- Data- ink ratio = $\frac{data ink}{total ink used to print the graphic}$
- Data ink ratio is the proportion of a graphic's ink devoted to the non-redundant display of data information
- Data ink ratio = 1.0 - proportion of the graphic that can be erased without loss of information
- The larger the share of graphic's ink devoted to data, the better: Maximise the data ink ratio within reason
- Every bit of ink on a graphic requires a reason
- Nearly always that reason should be that the ink presents new information
- Other principles bearing on graphic design follows the idea of maximising the share of data ink

**Two Erasing Principles**
- Erase non-data ink, within reason
- Unless redundancy has a worthy purpose: Erase redundant data ink, within reason
- Above all else show the data
- Maximise the data ink ratio
- Erase non-data ink
- Erase redundant data ink
- Revise and edit

**Chartjunk: Vibrations, Grids, and Ducks**
- Unintentional Optical Art
- Móire vibration is an undisciplined ambiguity, with an illusive, eye-straining quality that contaminates the entire graphic. It has no place in graphic design
- One of the most sedate graphical elements, the grid should usually be muted or completely suppressed so that its presence is only implicit, lest it compete with the data
- Dark grid lines are chart junk
- When a graphic serves as a look up table, then a grid may help in reading and interpolating. Even in this case grids should be muted relative to the data
- Most ready-made graph paper comes with a darkly printed grid. The reverse (unprinted side) should be used for then the lines show through faintly and don't clutter the data. If the paper is heavily grided on each side, throw it out!
- Forgo chartjunk, including móire vibration, the grid and the duck 

**Small Multiples**

- Small multiples resemble the frames of a movie: a series of graphics showing the same combination of variables, indexed by changes in another variable
- The design remains constant through all the frames, so the attention is devoted entirely to shifts in the data
- Well-designed small multiples are:
- Inevitably comparative
- Deftly multivariate
- Shrunken, high-density graphics
- Usually based on a large data matrix
- Drawn almost entirely with data ink
- Efficient in interpretation

- Often narrative in context, showing shifts in relationship between variables as the index variable changes
