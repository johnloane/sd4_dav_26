**Clustering**
- Machine learning technique that uses unsupervised labels
- Group unlabelled data into natural clusters
- Clustering is optimisation problem
- $variability(c) = \sum_{e \in c}distance(mean(c), e)^2$
- Variance is different from variability, you don't divide by n. This is because in clustering big and bad is worse than small and bad. We don't want to average out the difference
- $dissimilarity(C) = \sum_{c \in C}variability(c)$
- Find C that minimises the dissimilarity
- In order for this not end up as each point being a cluster, we add a constraint e.g the clusters have to have a minimum distance or constrain the number of clusters

**Hierarchical clustering**
- Start by assigning each item to a cluster so that if you have N items, you now have N clusters, each containing one item
- Find the closest (most similar) pair of clusters and merge them into a single cluster, so now we have one fewer clusters
- Continue until all items are clustered in a single cluster of size N
- Agglomerative hierarchical clustering
- What do we mean by closest?
- Can get clusters depending on the linkage metrics
- **Single linkage**: consider the distance between one cluster and another to be equal to shortest distance between any member of one cluster and any member of the other cluster
- **Complete linkage**: consider the distance between one cluster and another as the greatest from any member of one cluster to any member of the other cluster
- **Average linkage**: consider the distance between one cluster and another as the average distance from any member of one cluster to any member of the other cluster

**Pros and cons of hierarchical**
- Get history of dendrograms
- Deterministic given a linkage metric
- Answer might not be optimal. Greedy algorithm. Making locally optimal decisions at each point that might not be globally optimal
- Flexible
- Really slow. Complexity $O(n^3)$. Not something to use with large datasets

**k-means clustering**
- k is the number of clusters that you want
- If you don't have any idea about how many clusters there are this can be problematic
- If you do know how many clusters there are this is a good choice as it is very fast


**k-means algorithm**
- Randomly choice sample as initial centroid
- while True:
- create k clusters by assigning each point to the closest centroid
- compute k new centroids by averaging the examples in each cluster
- if centroids don't change we are finished

**k means complexity**
- Complexity of one iteration
- Typically converges quite quickly
- Most commonly used clustering algorithm because of its speed


**Choosing k problems**
- Choosing wrong k can be problematic
- If k is non-sensical then your clustering will be con-sensical
- How do we choose k?
- Results can depend on the initial centroids
- k-means is non-deterministic
- If the initial centroids are poorly chosen it can take longer to converge or you can get a completely different answer

**Choosing k solutions**
- Choose k with a-priori knowledge of the application
- For example there are only 5 different types of bacteria
- Two types of people in the world - DkIT graduates and not
- Try different values of k and evaluate the quality of the results
- Run hierarchical clustering on a subset of the data
- Try and select good initial centroids distributed over the space
- Try multiple sets of random chosen centroids and choose the best result


**Mitigating dependence on initial centroids**
- Best = kmeans(points)
- for t in range(numTrials):
- C = kmeans(points)
- if dissimilarity(C) < dissimilarity(best)
- best = C
- return C
- Think of the centroid as the "centre of gravity" of a cluster


**Elbow method to find k**
- $SSE_1 = \sum_{i=0}^n dist(x_i - c_1)^2$
- $SSE = SSE_1 + \dots + SSE_n$

**DBSCAN**
- Density based spatial clustering of applications with noise
- Clusters are dense regions in the data space, separated by regions of lower density
- DBSCAN is based on an intuitive notion of "clusters" and "noise"
- The key idea os that each point of a cluster, the neighbourhood of a given radius has to contain a given number of points
- Partitioning methods (k-means) and hierarchical clustering work for spherical shaped or convex clusters. They are suitable for compact and well separated clusters. They are severely impacted by outliers and noise
- Real life data can contain arbitrary shapes and noise

**Parameters for DBSCAN**
- $\epsilon$ defines the neighbourhood around a point. If the distance between two points is less than or equal to $\epsilon$ then they are considered neighbours. If the $\epsilon$ value is chosen too small then a large amount of our data will considered outliers
- **MinPts** Minimum number of neighbours (data points) within $\epsilon$ radius. The larger the dataset, the larger the value of MinPts should be. As a general rule, MinPts can be derived from the number of dimensions in the dataset, D, as MinPts >= D+1. The minimum value must be at least 3.
- In this algorithm there are 3 types of datapoints
- **Core point:** A point is a core point if it has at least MinPts within $\epsilon$ of it
- **Border point**: Fewever than MinPts within $\epsilon$ but it is in the neighbourhood of a core point
- **Noise of outlier**: A point that is not a core point or a border point

**Steps for DBSCAN algorithm**
- Find all points within $\epsilon$ and identify the core points with greater than or equal to MinPts neighbours
- For each core point if it is not already assigned to a cluster assign to the same cluster as the core point
- Recursively find all connected points and assign to the same cluster as the core point
- Iterate through the remaining points. Add border points.
- Points that don't belong to any cluster are noise

**Pseudo code for DBSCAN**
- DBSCAN(dataset, eps, MinPts)
- C = 1
- for each unvisited points p in the dataset
- mark p as visited
- find # neighbours for p
- if |N| >= MinPts
- Add p to C
- Add boundary points to C
- What is left is outliers

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn import metrics
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
```

```python
X, y = make_blobs(n_samples = 300, centers=4, cluster_std=0.5, random_state=0)
```

```python
X
```

```python
db = DBSCAN(eps=0.8, min_samples=5).fit(X)
```

```python
db.labels_
```

```python
core_samples_mask = np.zeros_like(db.labels_, dtype=bool)
```

```python
core_samples_mask[db.core_sample_indices_] = True
```

```python
labels = db.labels_
```

```python
#Number of clusters in the labels, ignoring noise if it is present
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
```

```python
n_clusters
```

```python
#Plot results
#Black to indicate noise
unique_labels = set(labels)
colours = ['y', 'b', 'g', 'r']
for k, col in zip(unique_labels, colours):
    if k == -1:
        #Black as it is noise
        col = 'k'
        
    class_member_mask = (labels == k)
    xy = X[class_member_mask & core_samples_mask]
    plt.plot(xy[:, 0], xy[:, 1], 'o', markerfacecolor=col, markeredgecolor='k', markersize = 6)
    xy = X[class_member_mask & ~core_samples_mask]
    plt.plot(xy[:, 0], xy[:, 1], 'x', markerfacecolor=col, markeredgecolor='k', markersize = 6)
```

**Evaluation metrics for DBSCAN**
- Silhouette score is in the range from -1 to 1. A score of 1 indicates that we have compact clusters. The worst value is -1. Value near 0 indicate overlapping clusters.
- Adjusted Rand Score is in the range from 0 to 1. More 0.9 denotes an excellent cluster, above 0.8 is good. Less than 0.5 is poor.

```python
sc = metrics.silhouette_score(X, labels)
sc
```

```python
from sklearn.metrics import adjusted_rand_score
ars = adjusted_rand_score(y, labels)
ars
```

|DBSCAN| k-means|
|---|---|
|In DBSCAN we don't need to specify the number of clusters|k-means is very sensitive to the number of clusters so k needs to be specified|
|Clusters formed in DBSCAN can be any arbitrary shape| Clusters formed in k-means are spherical or convex in shape|
|DBSCAN can work well with datasets that have noise or outliers|k_means does not work well with outliers. Outliers can skew k-means massively|
|In DBSCAN two parameters are required for training|In k-means we only need one parameter k|
|DBSCAN has higher complexity than k-means|

```python
# K-means works really poorly here, DBSCAN works well. Why?
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN

#Generate crescent moon shaped data
X, _ = make_moons(n_samples=300, noise=0.05, random_state=42)

#Run k-means
km = KMeans(n_clusters=2, random_state=42).fit(X)

#Run DBSCAN
db = DBSCAN(eps=0.2, min_samples=5).fit(X)

#Plot the results
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.scatter(X[:, 0], X[:, 1], c = km.labels_, cmap='viridis')
ax1.set_title("K-Means: Fails to capture the Moon shapes")

ax2.scatter(X[:, 0], X[:, 1], c = db.labels_, cmap='viridis')
ax2.set_title("DBSCAN: Correctly identifies the Moon shapes")
```
